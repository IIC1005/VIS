## 📊 Small Multiple

Un **Small Multiple** es una matriz de gráficos del mismo tipo, uno por cada subconjunto de los datos.  
Sirve para comparar patrones visuales de forma paralela y ordenada.

### 🧰 ¿Cómo se implementa con Seaborn?

Utilizamos la función `FacetGrid` para crear una **grilla de gráficos**. Esta recibe:

- `data`: el dataset completo,  
- `col`: la columna que define cómo se separan los gráficos,  
- `col_wrap`: para definir cuántos gráficos por fila se muestran.

Luego, usamos `.map()` para definir qué tipo de gráfico dibujar en cada celda y qué ejes usar (`x`, `y`).

### 🔗 Más información:
